```python
titanic["family_size"] = titanic["SibSp"]+titanic["Parch"]+1
```

```python
titanic["Fareperperson"] = titanic["Fare"]/titanic["family_size"]
```

```python
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
num_pipeline = Pipeline([
                    ("imputer",SimpleImputer(strategy="mean")),
                     ("scaler",StandardScaler())
                     ])
num_cat = ["Age","SibSp","Parch","Fare","family_size","Fareperperson"]
cat_1 = ["Sex","Embarked"]
cat_2 = ["Pclass"]

```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import OrdinalEncoder
num_cat = ["Age","SibSp","Parch","Fare","family_size","Fareperperson"]
cat_1 = ["Sex","Embarked"]
cat_2 = ["Pclass"]
full_pipeline = ColumnTransformer([("imputer",num_pipeline,num_cat),
                                  ("cat1",OneHotEncoder(),cat_1),
                                  ("cat2",OrdinalEncoder(),cat_2)])
full_pipeline.fit_transform(titanic)
```

```python
import numpy as np
from zlib import crc32 
def test_set(id,test_ratio):
    return crc32(np.int64(id)) & 0xffffffff < test_ratio*2**32
def split(data,id_col,test_ratio):
    id1 = data[id_col]
    in_test_set = id1.apply(lambda id2 : test_set(id2,test_ratio))
    return data.loc[in_test_set],data.loc[~in_test_set]

    
    
```

```python
titanic.head()
```

```python
test_set ,train_set = split(titanic,"PassengerId" ,0.2)
```

```python
from sklearn.linear_model import LogisticRegression
model = LogisticRegression()
x_train = full_pipeline.fit_transform(train_set_x)
model.fit(x_train,train_set_y)
x_test = full_pipeline.transform(test_set_x)
pred = model.predict(x_test)
```

```python
from sklearn.metrics import accuracy_score
acu_lr = accuracy_score(test_set_y,pred)
print(acu_lr)
```
